```python
# 1 & 2. Import all the necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from google.colab import files
```

```python
# 1. First, import the dataset in your google colab notebook
print("Please upload your dataset file:")
uploaded = files.upload()
```

```python
# Get the file name automatically from the upload
file_name = list(uploaded.keys())[0]
```

```python
# Read the data using pandas (no header row in this data)
df = pd.read_csv(file_name, header=None)
X = df[1].values
Y = df[2].values
```

```python
# 3. Apply the regression formula using Gradient Descent equations
m = 0  # Initial slope
c = 0  # Initial intercept
L = 0.0001  # Learning rate
epochs = 1000  # Number of iterations
n = float(len(X))
```

```python
# Gradient Descent Loop
for i in range(epochs):
    Y_pred = m * X + c
```

```python
# Calculate derivatives (gradients)
D_m = (-2/n) * sum(X * (Y - Y_pred))
D_c = (-2/n) * sum(Y - Y_pred)

# Update parameters
m = m - L * D_m
c = c - L * D_c
```

```python
print(f"\nFinal calculated slope (m): {m}")
print(f"Final calculated intercept (c): {c}")
```

```python
# 4. Plot the regression graph for the given dataset
plt.figure(figsize=(8, 5))
plt.scatter(X, Y, color='blue', label='Actual Data Points')
plt.plot([min(X), max(X)], [min(X)*m + c, max(X)*m + c], color='red', linewidth=2, label='Gradient Descent Line')
plt.title('Full Implementation of Regression Analysis')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.legend()
plt.grid(True)
plt.show()
```
